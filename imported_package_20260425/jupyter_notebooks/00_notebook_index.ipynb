# Teaching Notebooks

这个目录按实际使用顺序分成两组：先是巡线模型，再是 YOLO。

## 巡线模型

1. `01_line_follower_train_and_convert_workflow.ipynb`：从 `/home/ubuntu/数据` 训练 ResNet18 XY，并导出 ONNX / bin
2. `02_line_follower_bin_convert_workflow.ipynb`：已有 `.pth` 或 `.onnx` 时直接转 bin；也会自动同步 `/home/ubuntu/数据`

## YOLO

3. `03_yolo_train_workflow.ipynb`：YOLO11 训练
4. `04_yolo_bin_convert_workflow.ipynb`：YOLO ONNX 模型转 bin

如果你现在做巡线模型，直接从 01 开始按顺序看。
如果你已经有巡线模型权重或 ONNX，只需要打开 02。
